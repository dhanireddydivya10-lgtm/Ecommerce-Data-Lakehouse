```python
source_path = "/Volumes/workspace/default/lakehouse_volume/raw_data/"

schema_path = "/Volumes/workspace/default/lakehouse_volume/bronze_schema/"

checkpoint_path = "/Volumes/workspace/default/lakehouse_volume/bronze_checkpoint_v2/"

bronze_df = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option("header", "true")
    .option("cloudFiles.schemaLocation", schema_path)
    .option("cloudFiles.inferColumnTypes", "false")
    .load(source_path)
)

# Rename column names
for old_name in bronze_df.columns:
    new_name = old_name.strip().lower().replace(" ", "_")
    bronze_df = bronze_df.withColumnRenamed(old_name, new_name)

# Write into Bronze Delta table
(
    bronze_df.writeStream
    .format("delta")
    .option("checkpointLocation", checkpoint_path)
    .trigger(availableNow=True)
    .toTable("workspace.default.bronze_products")
)

print("Bronze Layer Created Successfully!")
```

```python
from pyspark.sql.functions import col, trim, when, expr

silver_df = spark.table("workspace.default.bronze_products")

# Clean spaces and empty strings
for c in silver_df.columns:
    silver_df = silver_df.withColumn(
        c,
        when(trim(col(c)) == "", None).otherwise(trim(col(c)))
    )

# Remove duplicate rows
silver_df = silver_df.dropDuplicates()

# Safely convert numeric columns
numeric_columns = {
    "list_price": "DOUBLE",
    "sale_price": "DOUBLE",
    "discount_percentage": "DOUBLE",
    "average_rating": "DOUBLE",
    "number_of_ratings": "INT"
}

for column_name, data_type in numeric_columns.items():
    silver_df = silver_df.withColumn(
        column_name,
        expr(f"try_cast(`{column_name}` AS {data_type})")
    )

# Save Silver table
silver_df.write.format("delta").mode("overwrite").saveAsTable(
    "workspace.default.silver_products"
)

print("Silver Layer Created Successfully!")
print("Silver Records:", silver_df.count())

display(silver_df.limit(10))
```

```python
from pyspark.sql.functions import (
    col, count, avg, round, sum, when
)

silver_df = spark.table("workspace.default.silver_products")

gold_df = (
    silver_df
    .groupBy("brand")
    .agg(
        count("*").alias("total_products"),
        round(avg("list_price"), 2).alias("avg_list_price"),
        round(avg("sale_price"), 2).alias("avg_sale_price"),
        round(avg("discount_percentage"), 2).alias("avg_discount"),
        round(avg("average_rating"), 2).alias("avg_rating"),
        sum("number_of_ratings").alias("total_ratings")
    )
)

gold_df.write.format("delta").mode("overwrite").saveAsTable(
    "workspace.default.gold_brand_summary"
)

print("Gold Layer Created Successfully!")

display(gold_df)
```

```python
from pyspark.sql.functions import col, count, avg, round, sum, coalesce, lit

silver_df = spark.table("workspace.default.silver_products")

gold_df = (
    silver_df
    .withColumn("brand", coalesce(col("brand"), lit("Unknown")))
    .groupBy("brand")
    .agg(
        count("*").alias("total_products"),
        round(avg("list_price"), 2).alias("avg_list_price"),
        round(avg("sale_price"), 2).alias("avg_sale_price"),
        round(avg("discount_percentage"), 2).alias("avg_discount"),
        round(avg("average_rating"), 2).alias("avg_rating"),
        sum("number_of_ratings").alias("total_ratings")
    )
)

gold_df.write.format("delta").mode("overwrite").saveAsTable(
    "workspace.default.gold_brand_summary"
)

display(gold_df)
```

```python
from delta.tables import DeltaTable

target = DeltaTable.forName(
    spark,
    "workspace.default.silver_products"
)

source = spark.table(
    "workspace.default.silver_products"
).limit(1)

target.alias("target").merge(
    source.alias("source"),
    "target.skuid = source.skuid"
).whenMatchedUpdateAll().whenNotMatchedInsertAll().execute()

print("Delta MERGE completed successfully!")
```
